```python

import pandas as pd
from pathlib import Path

project_folder = Path.cwd().parent
csv_path = project_folder / "data" / "cleaned_churn_data.csv"

df = pd.read_csv(csv_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python

# Check missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Check total missing values
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python

# Count duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())
```

```python

# Display column names
print(df.columns.tolist())

# Find TotalCharges even if the column has a space
total_col = next(
    (col for col in df.columns
     if col.strip().lower().replace(" ", "") == "totalcharges"),
    None
)

if total_col is not None:
    print("\nColumn found:", total_col)
    print("Data type before conversion:", df[total_col].dtype)

    # Convert to numeric; invalid values become NaN
    df[total_col] = pd.to_numeric(
        df[total_col].astype(str).str.strip().replace("", pd.NA),
        errors="coerce"
    )

    print("Data type after conversion:", df[total_col].dtype)
    print("Missing values after conversion:", df[total_col].isna().sum())
else:
    print("\nTotalCharges column not found.")
```

```python

# Check missing values before handling
print("Missing values before handling:")
print(df.isnull().sum())

# Remove rows only if the Churn target is missing
churn_col = next(
    (col for col in df.columns if col.strip().lower() == "churn"),
    None
)

if churn_col is not None:
    df = df.dropna(subset=[churn_col])

# Fill missing numeric values with the median
# Fill missing categorical values with the most frequent value
for col in df.columns:
    if col == churn_col:
        continue

    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            median_value = df[col].median()
            if pd.notna(median_value):
                df[col] = df[col].fillna(median_value)
        else:
            mode_values = df[col].mode(dropna=True)
            if not mode_values.empty:
                df[col] = df[col].fillna(mode_values.iloc[0])

print("\nMissing values after handling:")
print(df.isnull().sum())
```

```python

# Remove exact duplicate rows
df = df.drop_duplicates()

# Save the processed dataset
output_path = project_folder / "data" / "processed_churn_data.csv"

df.to_csv(output_path, index=False)

print("Processed dataset saved successfully!")
print("Saved to:", output_path)
print("Final dataset shape:", df.shape)
print("Remaining missing values:", df.isnull().sum().sum())
print("Remaining duplicate rows:", df.duplicated().sum())
```
