```python

import pandas as pd

# Load the final encoded dataset
encoded_df = pd.read_csv("../data/encoded_churn_data.csv")

print("=" * 60)
print("       CUSTOMER CHURN - FEATURE ENCODING REPORT")
print("=" * 60)

print("\n1. DATASET INFORMATION")
print("-" * 40)
print("Total rows:", encoded_df.shape[0])
print("Total columns:", encoded_df.shape[1])

print("\n2. ALL ENCODED COLUMN NAMES")
print("-" * 40)
for i, col in enumerate(encoded_df.columns, start=1):
    print(f"{i}. {col}")

print("\n3. FIRST FIVE ROWS")
print("-" * 40)
print(encoded_df.head().to_string(index=False))

print("\n4. MISSING VALUES")
print("-" * 40)
print("Total missing values:", encoded_df.isnull().sum().sum())

print("\n5. REMAINING TEXT COLUMNS")
print("-" * 40)
text_cols = encoded_df.select_dtypes(
    include=["object", "str"]
).columns.tolist()
print(text_cols if text_cols else "None")

print("\n6. CHURN LABEL DISTRIBUTION")
print("-" * 40)
print(encoded_df["Churn Label"].value_counts().sort_index())

print("\n7. FINAL STATUS")
print("-" * 40)
if (
    encoded_df.isnull().sum().sum() == 0
    and len(text_cols) == 0
    and encoded_df.shape == (7043, 34)
):
    print("Feature encoding completed successfully!")
    print("Dataset is ready for the next project stage.")
else:
    print("Please review the dataset before proceeding.")

print("=" * 60)
```

```python

# Reload the original cleaned dataset
df = pd.read_csv("../data/cleaned_churn_data.csv")

# Convert Total Charges to numeric
df['Total Charges'] = pd.to_numeric(
    df['Total Charges'], errors='coerce'
)

# Remove columns that aren't needed for this encoding task
df = df.drop(columns=['City', 'Lat Long', 'Zip Code'])

print("Dataset shape:", df.shape)
print(df.head())
```

```python

# Remove high-cardinality geographical columns
df = df.drop(
    columns=['City', 'Lat Long'],
    errors='ignore'
)

print("Dataset shape after removing columns:", df.shape)

print("\nRemaining categorical columns:")
print(df.select_dtypes(include=['object', 'str']).columns.tolist())
```

```python

# Select remaining categorical columns, excluding the target
remaining_cols = [
    col for col in df.select_dtypes(include=['object', 'str']).columns
    if col != 'Churn Label'
]

print("Columns to encode:", remaining_cols)

# Apply One-Hot Encoding
df = pd.get_dummies(
    df,
    columns=remaining_cols,
    drop_first=True,
    dtype=int
)

print("\nEncoding completed!")
print("New dataset shape:", df.shape)
print("\nFirst five rows:")
print(df.head())
```

```python

from sklearn.preprocessing import LabelEncoder

# Binary categorical columns
binary_cols = [
    'Gender',
    'Senior Citizen',
    'Partner',
    'Dependents',
    'Phone Service',
    'Paperless Billing',
    'Churn Label'
]

# Apply Label Encoding
label_encoders = {}

for col in binary_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    label_encoders[col] = le

print("Binary columns encoded successfully!")

print("\nEncoded values:")
print(df[binary_cols].head())

print("\nData types:")
print(df[binary_cols].dtypes)
```
