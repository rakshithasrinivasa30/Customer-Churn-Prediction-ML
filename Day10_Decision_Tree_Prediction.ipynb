```python
import joblib

# Load the saved best Decision Tree model
best_dt_model = joblib.load("../models/best_decision_tree_model.pkl")

print("Best Decision Tree model loaded successfully!")
print("Number of features expected:", best_dt_model.n_features_in_)
```

```python

import pandas as pd

# Load sample customer records
customer_raw = pd.read_csv("../data/sample_customer_records.csv")

# Check required columns
required_columns = [
    'Latitude', 'Longitude', 'Tenure Months',
    'Monthly Charges', 'Total Charges', 'CLTV',
    'Gender', 'Senior Citizen', 'Partner', 'Dependents',
    'Phone Service', 'Multiple Lines', 'Internet Service',
    'Online Security', 'Online Backup', 'Device Protection',
    'Tech Support', 'Streaming TV', 'Streaming Movies',
    'Contract', 'Paperless Billing', 'Payment Method'
]

missing = [col for col in required_columns if col not in customer_raw.columns]

if missing:
    raise ValueError(f"Missing columns in sample CSV: {missing}")

# Select only the required input columns
customer_df = customer_raw[required_columns].copy()

# Convert Senior Citizen from 0/1 to Yes/No
customer_df['Senior Citizen'] = customer_df['Senior Citizen'].replace({
    0: 'No', 1: 'Yes', '0': 'No', '1': 'Yes'
})

# Categorical columns
categorical_cols = [
    'Gender', 'Senior Citizen', 'Partner', 'Dependents',
    'Phone Service', 'Multiple Lines', 'Internet Service',
    'Online Security', 'Online Backup', 'Device Protection',
    'Tech Support', 'Streaming TV', 'Streaming Movies',
    'Contract', 'Paperless Billing', 'Payment Method'
]

# Encode categorical values
customer_encoded = pd.get_dummies(
    customer_df,
    columns=categorical_cols
)

# Match the exact 33 features used during training
customer_encoded = customer_encoded.reindex(
    columns=best_dt_model.feature_names_in_,
    fill_value=0
)

# Predict churn
predictions = best_dt_model.predict(customer_encoded)

# Display results
for i, prediction in enumerate(predictions):
    result = "Churn" if prediction == 1 else "No Churn"
    print(f"Customer {i + 1}: {result}")
```
