```python

import pandas as pd
import matplotlib.pyplot as plt

results = {
    "Model": [
        "Original Logistic Regression",
        "Tuned Logistic Regression",
        "Original Decision Tree",
        "Tuned Decision Tree"
    ],
    "Accuracy": [0.7871, 0.8006, 0.7871, 0.7864],
    "Precision": [0.6135, 0.6288, 0.5907, 0.6130],
    "Recall": [0.5348, 0.5722, 0.6444, 0.5294],
    "F1-score": [0.5714, 0.6037, 0.6164, 0.5681]
}

comparison_df = pd.DataFrame(results)

comparison_df.style.format({
    "Accuracy": "{:.2%}",
    "Precision": "{:.2%}",
    "Recall": "{:.2%}",
    "F1-score": "{:.2%}"
})
```

```python

plt.figure(figsize=(11, 6))

plt.bar(
    comparison_df["Model"],
    comparison_df["F1-score"]
)

plt.title("F1-Score Comparison of Four Churn Prediction Models")
plt.xlabel("Model")
plt.ylabel("F1-score")

plt.ylim(0, 1)
plt.xticks(rotation=20, ha="right")
plt.tight_layout()
plt.show()
```

## Final Model Comparison and Conclusion

We compared four models for customer churn prediction: Original Logistic Regression, Tuned Logistic Regression, Original Decision Tree, and Tuned Decision Tree.

The models were evaluated using accuracy, precision, recall, and F1-score.

The Tuned Logistic Regression achieved the highest accuracy of 80.06%. However, the Original Decision Tree achieved the highest F1-score of 61.64% and the highest recall of 64.44%.

The results show that tuning did not improve every metric for both algorithms. In particular, the Tuned Decision Tree had lower recall and F1-score than the Original Decision Tree.

Based on the current results, we recommend the Original Decision Tree as a candidate for the final model because it achieved the highest F1-score and recall. The final selection should be confirmed after checking that all models were evaluated on the same test set and with the same evaluation procedure.
